```python
# feature construction - effective area* price per sq ft

(
    houses
    .assign(
        effective_area = lambda df:(
            np.where(
                df.super_area.isna(),
                df.carpet_area,
                df.super_area
            )
        ),
        effective_cost = lambda df:(
            (df.price)*(df.effective_area)/1e7
        )
    )
    .pipe(
        eda.num_bivar_plots,
        "effective_cost",
        "amount"
    )
)
```

```python
(
    houses[["price","carpet_area","super_area"]]
    .assign(
        effective_area = lambda df:(
            np.where(
                df.super_area.isna(),
                df.carpet_area,
                df.super_area
            )
        ),
        effective_cost = lambda df:(
            (df.price)*(df.effective_area)/1e7
        )
    )
    .pipe(
        eda.correlation_heatmap
    )
)
```
